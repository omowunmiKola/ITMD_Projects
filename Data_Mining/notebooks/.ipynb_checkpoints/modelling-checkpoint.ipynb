```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# 1. Group the column names based on your dictionary
nominal_cols = [col for col, rules in BRFSS_MAP.items() if rules['type'] == 'nominal' and not rules.get('is_target')]
numeric_cols = [col for col, rules in BRFSS_MAP.items() if rules['type'] in ['ordinal', 'continuous'] and not rules.get('is_target')]

# 2. Build the transformer for Rule 2 (Numeric)
# This imputes the median AND automatically creates the missing indicator columns
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median', add_indicator=True))
])

# 3. Build the transformer for Rule 1 (Nominal)
# This One-Hot Encodes all categories, including the 'Unknown' level we just made
nominal_transformer = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore', drop='first'))
])

# 4. Combine them into your master preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_cols),
        ('cat', nominal_transformer, nominal_cols)
    ])
```
