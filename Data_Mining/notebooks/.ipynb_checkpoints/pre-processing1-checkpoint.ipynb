```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings('ignore')

#import pyreadstat

#df, meta = pyreadstat.read_xport('../LLCP2025.XPT')
RAW = pd.read_sas("../../data/LLCP2025.XPT ", format="xport", encoding="latin-1")
RAW = RAW.replace(5.397605346934028e-79, 0)
print(RAW.head(5))

RAW = RAW.rename(columns=lambda x: x.strip())

cols = RAW.columns

SEED = 42

#print(meta.column_names)
```

```python
keep =['_STATE','_METSTAT','_URBSTAT','SEXVAR','_AGE80',
       '_RACE','MARITAL','EDUCA','RENTHOM1','VETERAN3',
       'EMPLOY1','CHILDREN','INCOME3','PREGNANT','GENHLTH',
       'PHYSHLTH','MENTHLTH','PRIMINS2','MEDCOST1','PERSDOC3',
       'CHECKUP1','EXERANY2','BPHIGH6','TOLDHI3','CVDINFR4',
       'CVDCRHD4','CVDSTRK3','ASTHMA3','CHCSCNC1','CHCOCNC1',
       'CHCCOPD3','ADDEPEV3','CHCKDNY2','HAVARTH4', 
       '_BMI5','DEAF','BLIND','DECIDE','DIFFWALK','DIFFDRES',
       'DIFFALON','SLEPTIM1','_SMOKER3','USENOW3','ECIGNOW3',
       'DRNKANY6','FLUSHOT7','HIVTST7','PNEUVAC4','TETANUS1',
       'FRUIT2','FRUITJU2','FVGREEN1','FRENCHF1','POTATOE1','VEGETAB2','DIABETE4',]
df = RAW[keep].copy()

# Show first 10 rows of all columns
print(df.head(10))
```

```python
df = df[df["DIABETE4"].isin([1, 3])]
df["CLASS"] = np.where(df["DIABETE4"]== 3, 0, 1)
df = df.drop(columns=['DIABETE4'])
df.head()
```

```python
# Stratified holdout for testing
from sklearn.model_selection import train_test_split
df_train, df_holdout = train_test_split(
    df, 
    test_size=0.20, 
    stratify=df[TARGET], 
    random_state= SEED # Keeps the split reproducible
)
```

```python
#dev subsample
df_dev, _ = train_test_split(
    df_train, 
    train_size=50000, 
    stratify=df_train[TARGET], 
    random_state= SEED
)
```

```python
# Save to parquet
df_train.to_parquet('../../data/train.parquet', index=False)
df_holdout.to_parquet('../../data/holdout.parquet', index=False)
df_dev.to_parquet('../../data/dev.parquet', index=False)

print("Files successfully saved to data/ directory.")
```

```python
cols_to_process = ['FRUIT2','FRUITJU2','FVGREEN1','FRENCHF1','POTATOE1','VEGETAB2']
new_names =       ['FRUTDA2_','FTJUDA2_','GRENDA1_','FRNCHDA_','POTADA1_','VEGEDA2_']

def decode_brfss_frequency(val):
    # Handle missing, "Don't know" (777), and "Refused" (999)
    if pd.isna(val) or val >= 700:
        return np.nan
    # 555 = Never
    elif val == 555:
        return 0.0
    # 300 = Less than once a month
    elif val == 300:
        return 0.02
    # 101-199 = Times per day
    elif 100 <= val < 200:
        return val - 100.0
    # 201-299 = Times per week 
    elif 200 < val < 300:
        return round((val - 200) / 7.0, 2)
    # 301-399 = Times per month 
    elif 300 < val < 400:
        return round((val - 300) / 30.0, 2)
    else:
        return np.nan

# Pair up the old and new column names using zip() and apply the function
for old_col, new_col in zip(cols_to_process, new_names):
    df[new_col] = df[old_col].apply(decode_brfss_frequency)

print(df[new_names].head())
```
